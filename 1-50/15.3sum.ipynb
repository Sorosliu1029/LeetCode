### 15. 3Sum

<p>Given an integer array nums, return all the triplets <code>[nums[i], nums[j], nums[k]]</code> such that <code>i != j</code>, <code>i != k</code>, and <code>j != k</code>, and <code>nums[i] + nums[j] + nums[k] == 0</code>.</p>

<p>Notice that the solution set must not contain duplicate triplets.</p>

<p>&nbsp;</p>
<p><strong class="example">Example 1:</strong></p>

<pre>
<strong>Input:</strong> nums = [-1,0,1,2,-1,-4]
<strong>Output:</strong> [[-1,-1,2],[-1,0,1]]
<strong>Explanation:</strong> 
nums[0] + nums[1] + nums[2] = (-1) + 0 + 1 = 0.
nums[1] + nums[2] + nums[4] = 0 + 1 + (-1) = 0.
nums[0] + nums[3] + nums[4] = (-1) + 2 + (-1) = 0.
The distinct triplets are [-1,0,1] and [-1,-1,2].
Notice that the order of the output and the order of the triplets does not matter.
</pre>

<p><strong class="example">Example 2:</strong></p>

<pre>
<strong>Input:</strong> nums = [0,1,1]
<strong>Output:</strong> []
<strong>Explanation:</strong> The only possible triplet does not sum up to 0.
</pre>

<p><strong class="example">Example 3:</strong></p>

<pre>
<strong>Input:</strong> nums = [0,0,0]
<strong>Output:</strong> [[0,0,0]]
<strong>Explanation:</strong> The only possible triplet sums up to 0.
</pre>

<p>&nbsp;</p>
<p><strong>Constraints:</strong></p>

<ul>
	<li><code>3 &lt;= nums.length &lt;= 3000</code></li>
	<li><code>-10<sup>5</sup> &lt;= nums[i] &lt;= 10<sup>5</sup></code></li>
</ul>

What's your idea?

borrow idea from two-sum problem
- use binary search to fast locate final element
- use frequency counter to fast forward to next possible element

---

```python
from collections import Counter
import bisect

class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        l = len(nums)
        nums.sort()
        frequency = Counter(nums)
        left = 0
        result = []
        
        def binary_search(target: int, lo: int) -> int:
            idx = bisect.bisect_left(nums, target, lo)
            if idx < l and nums[idx] == target:
                return idx
            return -1
            
        while left < l-2:
            right = left + 1
            while right < l-1:
                idx = binary_search(-nums[left]-nums[right], right+1)
                if idx >= 0:
                    result.append([nums[left], nums[right], nums[idx]])
                right += (frequency[nums[right]] - (0 if nums[left] != nums[right] else 1))

            left += frequency[nums[left]]

        return result
```

```python
s = Solution()
```

```python
s.threeSum([-1,0,1,2,-1,-4])
```

```python
s.threeSum([0,1,1])
```

```python
s.threeSum([0,0,0])
```
