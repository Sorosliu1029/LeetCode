### 5. Longest Palindromic Substring

<p>Given a string <code>s</code>, return <em>the longest</em> <span data-keyword="palindromic-string"><em>palindromic</em></span> <span data-keyword="substring-nonempty"><em>substring</em></span> in <code>s</code>.</p>

<pre>
<strong>Input:</strong> s = &quot;babad&quot;
<strong>Output:</strong> &quot;bab&quot;

</pre>

<p><strong class="example">Example 2:</strong></p>

<pre>

What's your idea?

DP, as hints show, consider `s[i+1,j-1], s[i,j-1], s[i+1,j]`

```python
# Wed, 03 Dec 2025 21:13:34 GMT version
# 🚀 Runtime: 9155 ms, Memory: 99.1 MB
# 🉑 Runtime Percentile: better than 5.00%, Memory Percentile: better than 5.88%
class Solution:
    def longestPalindrome(self, s: str) -> str:
        self.memory = {}
        self.dp(s, 0, len(s))
        max_seg_i, max_seg_j = max((k for k, v in self.memory.items() if v), key=lambda k: k[1]-k[0])
        return s[max_seg_i: max_seg_j]

    def dp(self, s, i, j):
        if i+1 > j:
            return False
        if (i,j) in self.memory:
            return self.memory[(i,j)]
        if i+1 == j:
            self.memory[(i,j)] = s[i]
            return s[i]

        if s[i] == s[j-1]:
            s_mid = self.dp(s, i+1, j-1)
            if s_mid:
                self.memory[(i,j)] = s_mid if s[i] == s_mid else True
                return self.memory[(i,j)]
            
        if s[j-1] == self.dp(s, i, j-1):
            self.memory[(i,j)] = s[j-1]
            return self.memory[(i,j)]
        if s[i] == self.dp(s, i+1, j):
            self.memory[(i,j)] = s[i]
            return self.memory[(i,j)]

        self.memory[(i,j)] = False
        return False
```

```python
# WA version for Input: "bananas"
class Solution:
    def longestPalindrome(self, s: str) -> str:
        dp = [(1, True)] * len(s)
        for i in range(1, len(s)):
            if s[i] == s[i-1] and dp[i-1][1]:
                    dp[i] = (dp[i-1][0]+1, True)
            else:
                symmetry_idx = i - 1 - dp[i-1][0]
                if symmetry_idx >= 0 and s[symmetry_idx] == s[i]:
                    dp[i] = (dp[i-1][0]+2, False)
                else:
                    dp[i] = (1, True)
        l, idx = max(((p[0], i) for i, p in enumerate(dp)))
        return s[idx+1-l:idx+1]
```

What's your idea?

DP, refer to Gemini (What a shame...)

$$ dp_{i,j} = 
    \begin{cases} 
        dp_{i+1,j-1} & \text{if } s_i = s_j \\
        False & \text{otherwise}
    \end{cases}
$$

---

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
        n = len(s)
        r = (0,0)
        dp = [[False] * n for _ in range(n)]
        for i in range(n):
            dp[i][i] = True
        for i in range(n-1):
            if s[i] == s[i+1]:
                dp[i][i+1] = True
                r = (i,i+1)
            
        for width in range(2, n):
            for i in range(n-width):
                j = i + width
                if s[i] == s[j] and dp[i+1][j-1]:
                    dp[i][j] = True
                    r = (i,j)
        i, j = r
        return s[i:j+1]
```
