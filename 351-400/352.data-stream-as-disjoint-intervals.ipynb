### 352. Data Stream as Disjoint Intervals

<p>Given a data stream input of non-negative integers <code>a<sub>1</sub>, a<sub>2</sub>, ..., a<sub>n</sub></code>, summarize the numbers seen so far as a list of disjoint intervals.</p>

<p>Implement the <code>SummaryRanges</code> class:</p>

<ul>
	<li><code>SummaryRanges()</code> Initializes the object with an empty stream.</li>
	<li><code>void addNum(int value)</code> Adds the integer <code>value</code> to the stream.</li>
	<li><code>int[][] getIntervals()</code> Returns a summary of the integers in the stream currently as a list of disjoint intervals <code>[start<sub>i</sub>, end<sub>i</sub>]</code>. The answer should be sorted by <code>start<sub>i</sub></code>.</li>
</ul>

data structure: disjoint set

```python
class SummaryRanges:

    def __init__(self):
        self.intervals = {}

    def __parent(self, value):
        while not isinstance(self.intervals[value], list):
            value = self.intervals[value]
        return value
        
    def addNum(self, value: int) -> None:
        if value in self.intervals:
            return
        self.intervals[value] = [value, value]
        if value-1 not in self.intervals and value+1 not in self.intervals:
            return
        if value-1 in self.intervals:
            parent = self.__parent(value-1)
            self.intervals[value] = parent
            self.intervals[parent][1] = value
        if value+1 in self.intervals:
            parent = self.__parent(value)
            value1_parent = self.__parent(value+1)
            self.intervals[parent][1] = self.intervals[value1_parent][1]
            self.intervals[value1_parent] = parent
            
    def getIntervals(self) -> list[list[int]]:
        intervals = [v for v in self.intervals.values() if isinstance(v, list)]
        return sorted(intervals)
```
