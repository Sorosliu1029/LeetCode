### 128. Longest Consecutive Sequence

<p>Given an unsorted array of integers <code>nums</code>, return <em>the length of the longest consecutive elements sequence.</em></p>

<p>You must write an algorithm that runs in&nbsp;<code>O(n)</code>&nbsp;time.</p>

<p>&nbsp;</p>
<p><strong class="example">Example 1:</strong></p>

<pre>
<strong>Input:</strong> nums = [100,4,200,1,3,2]
<strong>Output:</strong> 4
<strong>Explanation:</strong> The longest consecutive elements sequence is <code>[1, 2, 3, 4]</code>. Therefore its length is 4.
</pre>

<p><strong class="example">Example 2:</strong></p>

<pre>
<strong>Input:</strong> nums = [0,3,7,2,5,8,4,6,0,1]
<strong>Output:</strong> 9
</pre>

<p><strong class="example">Example 3:</strong></p>

<pre>
<strong>Input:</strong> nums = [1,0,1,2]
<strong>Output:</strong> 3
</pre>

<p>&nbsp;</p>
<p><strong>Constraints:</strong></p>

<ul>
	<li><code>0 &lt;= nums.length &lt;= 10<sup>5</sup></code></li>
	<li><code>-10<sup>9</sup> &lt;= nums[i] &lt;= 10<sup>9</sup></code></li>
</ul>

What's your idea?

Disjoint set union, start with individual sections, merge consecutive sections into larger one, while also flatten child nodes

---

```python
class Solution:
    def longestConsecutive(self, nums: list[int]) -> int:
        if not nums:
            return 0
            
        sections: dict[int, int | tuple[int, int]] = {} # n -> parent n or parent section

        def get_root(n) -> None | tuple[int, int]:
            sec = sections.get(n)
            parent = n
            while sec is not None and isinstance(sec, int):
                parent = sec
                sec = sections.get(parent)
            return sec, parent
            
        for n in nums:
            if n in sections:
                continue
                
            (left_sec, left_parent), (right_sec, right_parent) = get_root(n-1), get_root(n+1)
            if left_sec is not None and right_sec is not None:
                sections[n] = (left_sec[0], right_sec[1])
                sections[left_parent] = sections[right_parent] = n
            elif left_sec is not None:
                sections[n] = left_parent
                sections[left_parent] = (left_sec[0], n+1)
            elif right_sec is not None:
                sections[n] = right_parent
                sections[right_parent] = (n, right_sec[1])
            else:
                sections[n] = (n, n+1)

        return max((v[1]-v[0] for v in sections.values() if isinstance(v, tuple)))
```
